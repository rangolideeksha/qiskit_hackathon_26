```python
import numpy as np

ket0 = np.array([[1], [0]])
ket1 = np.array([[0], [1]])

print(ket0 / 2 + ket1 / 2)
```

```python
M1 = np.array([[1, 1], [0, 0]])
M2 = np.array([[1, 0], [0, 1]])
M = M1 / 2 + M2 / 2
print(M)
```

```python
print(M1 @ ket1)
print(M1 @ M2)
print(M @ M)
```

```python
from qiskit.visualization import array_to_latex

display(array_to_latex(M1 @ ket1))
display(array_to_latex(M1 @ M2))
display(array_to_latex(M @ M))
display(array_to_latex(M))
display(array_to_latex(M1))
display(array_to_latex(M2))
```

```python
from qiskit.quantum_info import Statevector
from numpy import sqrt

u = Statevector([1 / sqrt(2), 1 / sqrt(2)])
v = Statevector([(1 + 2.0j) / 3, -2 / 3])
w = Statevector([1 / 3, 2 / 3])

display(u.draw("latex"))
display(v.draw("latex"))
display(w.draw("latex"))
```

```python
outcome, state = v.measure()
print(f"Measured: {outcome}\nPost-measurement state:")
display(state.draw("latex"))
```

```python
from qiskit.visualization import plot_histogram

statistics = v.sample_counts(1000)
plot_histogram(statistics)
```

```python
from qiskit import QuantumCircuit

circuit = QuantumCircuit(1)

circuit.h(0)
circuit.t(0)
circuit.h(0)
circuit.s(0)
circuit.y(0)

display(circuit.draw(output="mpl"))
```

```python
zero = Statevector.from_label("0")
one = Statevector.from_label("1")
psi = zero.tensor(one)
display(zero.draw("latex"))
display(one.draw("latex"))
display(psi.draw("latex"))
```

```python
plus = Statevector.from_label("+")
minus_i = Statevector.from_label("l")
phi = plus.tensor(minus_i)
display(plus.draw("latex"))
display(minus_i.draw("latex"))
display(phi.draw("latex"))

display((plus ^ minus_i).draw("latex"))
```

```python
from qiskit.quantum_info import Operator
from numpy import sqrt

H = Operator.from_label("H")
Id = Operator.from_label("I")
X = Operator.from_label("X")
display(H.draw("latex"))
display(Id.draw("latex"))
display(X.draw("latex"))
display(H.tensor(Id).draw("latex"))
display(H.tensor(X).draw("latex"))
display(H.tensor(Id).tensor(X).draw("latex"))

display((H ^ Id ^ X).draw("latex"))
```

```python
w = Statevector([0, 1, 1, 0, 1, 0, 0, 0,] / sqrt(3))
display(w.draw("latex"))

result, state = w.measure([0])
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))

result, state = w.measure([0, 1])
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))
```
